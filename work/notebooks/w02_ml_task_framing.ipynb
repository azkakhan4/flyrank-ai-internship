### Target or Proxy

The target variable for my task is `is_declining_label`.

It is a binary target:
- 1 indicates that a content page is classified as declining.
- 0 indicates that a content page is not classified as declining.

This target is derived from the observed `trend_direction` column. It can be used as a proxy to identify pages that may need SEO attention.

The model will assign a score to each page based on its likelihood of belonging to the declining category. Pages with higher scores will receive higher priority for review.

**Important limitation:** This target represents observed historical decline, not guaranteed future performance.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

### Success Metric

The primary success metric for this task is Precision@K.

Precision@K measures how many of the top-ranked website pages are actually classified as declining.

For example, Precision@20 measures the proportion of correctly identified declining pages among the 20 highest-ranked pages.

This metric is suitable because the main goal is to help SEO specialists identify the most important pages to review first.

A higher Precision@K means fewer unnecessary reviews and better prioritization of content improvement efforts.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

## The Designated Explanation Cell:
### The Unit of Analysis

The unit of analysis is one website content page. Each row in the dataset represents an individual content page identified by a unique content_id.

The dataset contains 30,000 rows and 5 columns in our selected DataFrame. Each page includes information such as impressions, sessions, trend direction, and the declining label.

This page-level data helps us compare content performance and identify pages that may need SEO attention.

```python

import pandas as pd

# Load the anonymized starter dataset
url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

# Create the target label
df["is_declining_label"] = (
    df["trend_direction"] == "down"
).astype(int)

# Select the columns relevant to our task
lane_df = df[
    [
        "content_id",
        "impressions_90d",
        "sessions_90d",
        "trend_direction",
        "is_declining_label"
    ]
]

# Display the real dataframe
print("One row represents one website content page.")
print("Dataset shape:", lane_df.shape)

display(lane_df.head(10))

```
